# Explore the test file downloaded from satellite GOES16

```python
import xarray as xr
import plotly.express as px
```

```python
from src.co2sat.utils import project_root
```

```python
path = (
    project_root()
    / "data"
    / "raw"
    / "goes16_test"
    / "OR_ABI-L1b-RadC-M6C01_G16_s20210911201147_e20210911203520_c20210911203556.nc"
)
```

```python
ds = xr.open_dataset(path)
```

```python
print(ds)
```

```python
print("Variables:", list(ds.data_vars))
print("Number of variables:", len(ds.data_vars))
```

```python
print("Coordinates:", list(ds.coords))
```

```python
print("Radiance shape:", ds["Rad"].shape)
```

```python
print("\nProjection params:")
print(ds["goes_imager_projection"])
```

```python
print("First few x values in radians:", ds["x"].values[:5])
```

```python
print("First few y values in radians:", ds["y"].values[:5])
```

```python
fig = px.imshow(
    ds["Rad"],
    color_continuous_scale="Viridis",
    origin="lower",
    aspect="auto",
)

fig.update_layout(
    title="Band 1 (Blue, 0.47 μm) radiance — raw fixed-grid coordinates",
    width=900,
    height=500,
)
path = project_root() / "figures"
fig.write_html(path / "radiance_band1.html")
```
