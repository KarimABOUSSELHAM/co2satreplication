# Explore what a `Facility` file looks like in order download attributes later

```python
import os
import httpx
from dotenv import load_dotenv
import json
```

```python
load_dotenv()
api_key = os.environ.get("API_KEY")
base = os.environ.get("BASE")
```

```python
headers = {"x-api-key": api_key}
r = httpx.get(base, headers=headers, timeout=60)
items = r.json()["items"]
```

```python
# Filter to Facility files
facility_items = [it for it in items if it["metadata"].get("dataType") == "Facility"]
print(f"Total Facility files: {len(facility_items)}")
```

```python
for it in facility_items[:5]:
    print(json.dumps(it, indent=2))
    print("---")
```

It seems EPA's Facility files are ordered one per year (covering all states in one file) rather than one per state-year.

The metadata fields are `description`, `year` and `dataType`
